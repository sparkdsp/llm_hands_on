```python
import torch 
import tiktoken
import importlib
```

```python
print(f"tiktoken version: {importlib.metadata.version("tiktoken")}")
```

```python
text = (
    "This is an example, <|endoftext|>"
)
text_set = {
    "This is an example, <|endoftext|>"
}
print(type(text))
print(type(text_set))
```

```python
# 토큰 인코딩을 위한 tokenizer를 불러온다. 
tiktokenizer( = tiktoken.get_encoding("gpt2")
```

```python
# 예상 문제 1번, 텍스트를 토크나이저로 인코딩한다. 
text_token = tiktokenizer.encode(text, allowed_special={"<|endoftext|>"})
```

```python
text_token = tiktokenizer.encode(text, allowed_special={"<|endoftext|>"})
```

```python
type(text_token)
```

```python
print(f"Token information is: {text_token}")
```

```python
# 인코딩 한 뒤에는 뭐를 해야할까? 
# 토큰을 텐서로 바꾸자. 
text_tensor = torch.tensor(text_token)
```

```python
print(f"Text tensor information is: {text_tensor}")
```

```python
# 스트링으로 받아서, 인코딩 하고, 리스트로 생성된 것을 텐서로 바꾼다.
```

```python
# 예상 문제 2, token decoding
decoded_string = tiktokenizer.decode(text_token)
```

```python
print(f"Decoded String is: {decoded_string}")
```
